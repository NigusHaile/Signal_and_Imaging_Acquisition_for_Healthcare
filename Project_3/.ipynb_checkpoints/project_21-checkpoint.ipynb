# BUSI → PyRadiomics → BI-RADS-aligned LLM-ready Features

Pipeline (matches the EXTRACT + FUSE blocks of the flow diagram):

  1. **Load BUSI** → discover image/mask pairs, merge multi-mask cases.
  2. **Validate masks** → QC (size, edge contact, area fraction).
  3. **Augment** (geometric only — preserves radiomic validity).
  4. **Extract radiomic features** — shape, first-order, GLCM, GLRLM, GLSZM, GLDM, NGTDM
     across filter banks (Original, Wavelet, LoG, Gradient, Square, SquareRoot,
     Logarithm, Exponential) → ~1000+ features per lesion.
  5. **BI-RADS lexicon mapping** — group features by the US descriptors they express
     (Shape, Orientation, Margin, Echo pattern, Posterior features).
  6. **Feature selection** (variance → correlation → LASSO) → ~20–50 stable features.
  7. **Serialize for LLM** — structured BI-RADS-style text report + compact JSON.

## Discover cases

```python
def list_cases(busi_dir):
    """Return list of {image, masks, class, case_id} dicts.
    
    Handles BUSI structure with separate image/mask folders:
        BUSI_dataset/benign/         ← images
        BUSI_dataset/benign_mask/    ← masks
        BUSI_dataset/malignant/
        BUSI_dataset/malignant_mask/
    """
    cases = []
    img_dirs = [d for d in busi_dir.iterdir() 
                if d.is_dir() and not d.name.endswith('_mask')]
    
    for img_dir in img_dirs:
        cls = img_dir.name
        mask_dir = busi_dir / f'{cls}_mask'
        if not mask_dir.exists():
            continue
        
        for img in sorted(img_dir.glob('*.png')):
            masks = sorted(mask_dir.glob(f'{img.stem}*.png'))
            if masks:
                cases.append({
                    'image': img, 'masks': masks,
                    'class': cls, 'case_id': img.stem
                })
    return cases

cases = list_cases(busi_dataset)
print(f'Total cases: {len(cases)}')

if cases:
    df = pd.DataFrame([{'class': c['class'], 'case_id': c['case_id'],
                        'n_masks': len(c['masks'])} for c in cases])
    print(df['class'].value_counts())
else:
    print('⚠ No cases found. Check filename matching between image and mask folders.')
```

## Load image + mask pair (with multi-mask merging)

```python
def load_pair(image_path, mask_paths):
    """Load image and merge multiple masks via logical OR."""
    img = np.array(Image.open(image_path).convert('L'), dtype=np.float32)
    merged = None
    for mp in mask_paths:
        m = (np.array(Image.open(mp).convert('L')) > 127).astype(np.uint8)
        merged = m if merged is None else (merged | m)
    return img, merged.astype(np.uint8)

# Show one example from each class
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for col, cls in enumerate(['benign', 'malignant', 'normal']):
    case = next((c for c in cases if c['class'] == cls), None)
    if not case: continue
    img, mask = load_pair(case['image'], case['masks'])
    axes[0, col].imshow(img, cmap='gray'); axes[0, col].set_title(f'{cls}: {case["case_id"]}')
    axes[1, col].imshow(mask, cmap='gray'); axes[1, col].set_title(f'mask ({mask.sum()} px)')
    #for ax in axes[:, col]: ax.axis('off')
plt.tight_layout(); plt.show()
```

## Mask validation

```python
def validate_mask(image, mask, min_voxels=100, max_area_frac=0.5):
    """Quick QC. Returns (ok: bool, info: dict)."""
    info = {'voxels': int(mask.sum()), 'area_frac': float(mask.sum() / mask.size)}
    if image.shape != mask.shape:
        return False, {**info, 'reason': f'shape mismatch {image.shape} vs {mask.shape}'}
    if info['voxels'] == 0:
        return False, {**info, 'reason': 'empty mask'}
    if info['voxels'] < min_voxels:
        return False, {**info, 'reason': f'too small ({info["voxels"]} < {min_voxels})'}
    if info['area_frac'] > max_area_frac:
        return False, {**info, 'reason': f'too large ({info["area_frac"]:.0%})'}
    edge = mask[0].sum() + mask[-1].sum() + mask[:, 0].sum() + mask[:, -1].sum()
    info['edge_pixels'] = int(edge)
    return True, info

# Validate all non-normal cases
results = []
for c in cases:
    if c['class'] == 'normal': continue
    img, mask = load_pair(c['image'], c['masks'])
    ok, info = validate_mask(img, mask)
    results.append({'case_id': c['case_id'], 'class': c['class'], 'ok': ok, **info})

val_df = pd.DataFrame(results)
print(f'Valid: {val_df["ok"].sum()} / {len(val_df)}')
val_df.head(10)
```

## PyRadiomics-safe augmentation

Only **geometric** transforms — preserves radiomic feature validity. Avoids brightness/blur/noise/elastic (which would corrupt texture features).

```python
def _rotate(arr, angle_deg, is_mask):
    s = sitk.GetImageFromArray(arr.astype(np.float32))
    c = [(sz - 1) / 2 for sz in s.GetSize()]
    t = sitk.Euler2DTransform(); t.SetCenter(s.TransformContinuousIndexToPhysicalPoint(c))
    t.SetAngle(np.deg2rad(angle_deg))
    interp = sitk.sitkNearestNeighbor if is_mask else sitk.sitkLinear
    out = sitk.GetArrayFromImage(sitk.Resample(s, s, t, interp, 0.0, s.GetPixelID()))
    return out.astype(np.uint8) if is_mask else out

def augment(image, mask, seed=42):
    """Yields (image, mask, tag) — original + flip + small rotation + mask perturbation."""
    rng = np.random.default_rng(seed)
    yield image, mask, 'original'
    yield np.fliplr(image).copy(), np.fliplr(mask).copy(), 'hflip'
    angle = float(rng.uniform(-10, 10))
    yield _rotate(image, angle, False), _rotate(mask, angle, True), f'rot{angle:+.1f}'
    # Mask perturbation (dilate or erode by 1 px)
    sm = sitk.GetImageFromArray(mask)
    op = sitk.BinaryDilateImageFilter() if rng.random() < 0.5 else sitk.BinaryErodeImageFilter()
    op.SetKernelRadius(1); op.SetForegroundValue(1)
    perturbed = sitk.GetArrayFromImage(op.Execute(sm)).astype(np.uint8)
    yield image.copy(), perturbed, 'mask_perturb'

# Demo on one case
sample = next(c for c in cases if c['class'] == 'malignant')
img, mask = load_pair(sample['image'], sample['masks'])
variants = list(augment(img, mask))

fig, axes = plt.subplots(1, len(variants), figsize=(14, 4))
for ax, (v_img, v_mask, tag) in zip(axes, variants):
    ax.imshow(v_img, cmap='gray')
    ax.contour(v_mask, levels=[0.5], colors='red', linewidths=1.5)
    ax.set_title(f'{tag}\n{v_mask.sum()} px'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## PyRadiomics feature extractor

```python
# Broader filter bank + explicit BI-RADS-relevant feature classes.
# Filters chosen per the flow diagram's EXTRACT box: wavelet · LoG · square · gradient.
# Adds SquareRoot, Logarithm, Exponential for additional intensity remappings, and
# tunes LoG sigmas to span fine (margin/spiculation) → coarse (overall heterogeneity).

def build_extractor():
    ext = featureextractor.RadiomicsFeatureExtractor(
        binWidth=25, normalize=True, normalizeScale=100,
        force2D=True, force2Ddimension=0, label=1,
        minimumROIDimensions=2, minimumROISize=50,
        interpolator=sitk.sitkBSpline,
        resampledPixelSpacing=None,
        geometryTolerance=1e-4,
    )

    # All standard feature classes. Each maps to BI-RADS US descriptors:
    #   shape        → Shape (oval/round/irregular), Orientation (parallel/not-parallel)
    #   firstorder   → Echo pattern (anechoic/hypo/iso/hyper/heterogeneous)
    #   glcm/glrlm   → Margin (circumscribed vs indistinct/angular/spiculated)
    #   glszm/gldm   → Heterogeneity, microlobulation, complex-cystic-and-solid
    #   ngtdm        → Coarseness / busyness (texture roughness)
    ext.disableAllFeatures()
    for fc in ['shape2D', 'firstorder', 'glcm', 'glrlm', 'glszm', 'gldm', 'ngtdm']:
        ext.enableFeatureClassByName(fc)

    # Filter bank — multi-scale views of the lesion (margin detail, edge sharpness).
    ext.disableAllImageTypes()
    ext.enableImageTypeByName('Original')
    ext.enableImageTypeByName('Wavelet')                            
    ext.enableImageTypeByName('LoG', customArgs={'sigma': [0.5, 1.0, 2.0, 3.0, 5.0]})
    ext.enableImageTypeByName('Gradient')
    ext.enableImageTypeByName('Square')
    ext.enableImageTypeByName('SquareRoot')
    ext.enableImageTypeByName('Logarithm')
    ext.enableImageTypeByName('Exponential')
    return ext


def posterior_features(image, mask):
    """BI-RADS US: posterior features (no posterior features / enhancement / shadowing).
    
    Not in PyRadiomics — derived directly from image+mask geometry: compare mean
    intensity in a band BELOW the lesion vs a same-area lateral band.
        ratio > 1   → posterior enhancement (brighter behind lesion)
        ratio < 1   → posterior shadowing
        ratio ≈ 1   → no posterior features
    """
    ys, xs = np.where(mask > 0)
    if len(ys) == 0:
        return {'posterior_ratio': float('nan'),
                'posterior_label': 'undefined',
                'posterior_strength': float('nan')}
    y_min, y_max = ys.min(), ys.max()
    x_min, x_max = xs.min(), xs.max()
    h, w = mask.shape
    lesion_h = y_max - y_min + 1
    lesion_w = x_max - x_min + 1

    # Band directly below the lesion, same width, height = 0.5 × lesion height.
    band_h = max(5, lesion_h // 2)
    pb_y0, pb_y1 = y_max + 1, min(h, y_max + 1 + band_h)
    pb_x0, pb_x1 = x_min, x_max + 1
    posterior_band = image[pb_y0:pb_y1, pb_x0:pb_x1]

    # Lateral reference band at the same depth (same y-range, beside the lesion).
    lat_w = lesion_w
    if x_min - lat_w >= 0:
        ref_x0, ref_x1 = x_min - lat_w, x_min
    elif x_max + 1 + lat_w <= w:
        ref_x0, ref_x1 = x_max + 1, x_max + 1 + lat_w
    else:
        ref_x0, ref_x1 = 0, min(w, lat_w)
    lateral_band = image[pb_y0:pb_y1, ref_x0:ref_x1]

    if posterior_band.size == 0 or lateral_band.size == 0:
        return {'posterior_ratio': float('nan'),
                'posterior_label': 'undefined',
                'posterior_strength': float('nan')}

    pm = float(posterior_band.mean())
    lm = float(lateral_band.mean()) + 1e-6
    ratio = pm / lm
    if ratio > 1.15:    label = 'enhancement'
    elif ratio < 0.85:  label = 'shadowing'
    else:               label = 'no_posterior_features'
    return {'posterior_ratio': ratio,
            'posterior_label': label,
            'posterior_strength': abs(ratio - 1.0)}


def orientation_feature(mask):
    """BI-RADS US: orientation (parallel vs not-parallel to skin line).
    Heuristic: ratio of bounding-box width to height. >1 ⇒ parallel (US convention).
    """
    ys, xs = np.where(mask > 0)
    if len(ys) == 0:
        return {'orientation_wh_ratio': float('nan'), 'orientation_label': 'undefined'}
    w = float(xs.max() - xs.min() + 1)
    h = float(ys.max() - ys.min() + 1)
    ratio = w / max(h, 1.0)
    label = 'parallel' if ratio >= 1.0 else 'not_parallel'
    return {'orientation_wh_ratio': ratio, 'orientation_label': label}


def extract_features(image, mask, ext):
    s_img = sitk.GetImageFromArray(image.astype(np.float32))
    s_mask = sitk.GetImageFromArray(mask.astype(np.uint8))
    s_mask.CopyInformation(s_img)
    result = ext.execute(s_img, s_mask)
    feats = {k: float(v) for k, v in result.items()
             if not k.startswith('diagnostics_')
             and isinstance(v, (int, float, np.number))}
    # Augment with BI-RADS-specific descriptors not covered by PyRadiomics.
    feats.update(posterior_features(image, mask))
    feats.update(orientation_feature(mask))
    return feats


extractor = build_extractor()
feats = extract_features(img, mask, extractor)
numeric_feats = {k: v for k, v in feats.items() if isinstance(v, (int, float))}
print(f'Extracted {len(feats)} features '
      f'({len(numeric_feats)} numeric, {len(feats) - len(numeric_feats)} categorical/label)')
pd.Series({k: v for k, v in numeric_feats.items()}).head(10)
```

## Group features by BI-RADS-relevant category

```python
# BI-RADS US lexicon → radiomic-feature mapping.
# radiomic features whose values most directly express it.
#
# Reference: ACR BI-RADS Atlas 5th ed. (Quick Reference, ULTRASOUND section).
#   Shape         : Oval / Round / Irregular
#   Orientation   : Parallel / Not parallel
#   Margin        : Circumscribed / Indistinct / Angular / Microlobulated / Spiculated
#   Echo pattern  : Anechoic / Hypo / Iso / Hyper / Complex / Heterogeneous
#   Posterior     : None / Enhancement / Shadowing / Combined
#   Associated    : Architectural distortion, duct changes, etc. (partly covered)

BIRADS_LEXICON = {
    # SHAPE (BI-RADS: oval / round / irregular)
    'shape': {
        'descriptor': 'Shape',
        'options':    ['oval', 'round', 'irregular'],
        'features':   ['Sphericity', 'Elongation', 'MajorAxisLength',
                       'MinorAxisLength', 'PerimeterSurfaceRatio', 'PixelSurface'],
    },
    #  ORIENTATION (BI-RADS: parallel / not-parallel)
    'orientation': {
        'descriptor': 'Orientation',
        'options':    ['parallel', 'not_parallel'],
        'features':   ['orientation_wh_ratio', 'orientation_label'],
    },
    # MARGIN (BI-RADS: circumscribed / indistinct / angular / microlobulated / spiculated)
    # GLCM/GLRLM at the lesion edge + wavelet HH detail + LoG fine sigma capture margin sharpness.
    'margin_smooth': {
        'descriptor': 'Margin (circumscribed vs not)',
        'options':    ['circumscribed', 'not_circumscribed'],
        'features':   ['Sphericity', 'PerimeterSurfaceRatio',
                       'glcm_Idmn', 'glcm_Idn', 'glcm_Imc1'],
    },
    'margin_spiculation': {
        'descriptor': 'Margin — spiculation / indistinctness',
        'options':    ['indistinct', 'angular', 'spiculated'],
        'features':   ['wavelet-HH', 'wavelet-HL', 'wavelet-LH',
                       'log-sigma-0-5', 'log-sigma-1-0',
                       'gradient_'],
    },
    # ECHO PATTERN (BI-RADS: anechoic / hypo / iso / hyper / heterogeneous / complex)
    'echo_pattern': {
        'descriptor': 'Echo pattern',
        'options':    ['anechoic', 'hypoechoic', 'isoechoic',
                       'hyperechoic', 'complex_cystic_solid', 'heterogeneous'],
        'features':   ['firstorder_Mean', 'firstorder_Median',
                       'firstorder_Entropy', 'firstorder_Skewness',
                       'firstorder_Kurtosis', 'firstorder_Uniformity',
                       'firstorder_10Percentile', 'firstorder_90Percentile'],
    },
    # HETEROGENEITY / coarseness (refines the "heterogeneous" echo subtype)
    'heterogeneity': {
        'descriptor': 'Internal heterogeneity (refines echo pattern)',
        'options':    ['homogeneous', 'mildly_heterogeneous', 'markedly_heterogeneous'],
        'features':   ['glcm_Contrast', 'glcm_Correlation',
                       'glcm_JointEntropy', 'glcm_DifferenceEntropy',
                       'glrlm_GrayLevelNonUniformity', 'glszm_ZoneEntropy',
                       'gldm_DependenceEntropy', 'ngtdm_Coarseness',
                       'ngtdm_Busyness', 'ngtdm_Contrast'],
    },
    # POSTERIOR FEATURES (BI-RADS: none / enhancement / shadowing / combined)
    'posterior': {
        'descriptor': 'Posterior features',
        'options':    ['no_posterior_features', 'enhancement', 'shadowing', 'combined'],
        'features':   ['posterior_ratio', 'posterior_label', 'posterior_strength'],
    },
    #  MULTI-SCALE (filter bank, supports margin + texture descriptors)
    'multiscale': {
        'descriptor': 'Multi-scale wavelet / LoG signatures',
        'options':    ['(quantitative only)'],
        'features':   ['wavelet-LL', 'log-sigma-2-0', 'log-sigma-3-0', 'log-sigma-5-0'],
    },
}


def _match_any(key, patterns):
    return any(p.lower() in key.lower() for p in patterns)


def categorize_lexicon(features):
    """Group features by BI-RADS US descriptor (lexicon-aligned)."""
    out = {g: {} for g in BIRADS_LEXICON}
    out['other'] = {}
    for key, val in features.items():
        placed = False
        for group, spec in BIRADS_LEXICON.items():
            if _match_any(key, spec['features']):
                out[group][key] = val
                placed = True
                break
        if not placed:
            out['other'][key] = val
    return out


# Sanity check the mapping
grouped = categorize_lexicon(feats)
pd.Series({g: len(v) for g, v in grouped.items()}).sort_values(ascending=False)
```

## Feature selection  "1000+ radiomic features → 20–50 stable features"
#`outputs/busi_features_selected.csv`.

```python
def select_stable_features(feature_df, label_col='class', target_k=40,
                            var_thresh=1e-4, corr_thresh=0.92, lasso_C=0.1):
    """Three-stage selection on a (rows=samples, cols=features) DataFrame.
    1. Drop near-constant columns (variance < var_thresh after z-scoring).
    2. Drop one of each pair with |corr| > corr_thresh.
    3. L1-logistic regression (LASSO) keeps the top-`target_k` by |coefficient|.
    """
    from sklearn.preprocessing import StandardScaler
    from sklearn.linear_model import LogisticRegression
    from sklearn.feature_selection import VarianceThreshold

    meta_cols = [c for c in ['class', 'case_id', 'augmentation'] if c in feature_df.columns]
    X = feature_df.drop(columns=meta_cols).select_dtypes(include=[np.number]).copy()
    X = X.replace([np.inf, -np.inf], np.nan).fillna(X.median(numeric_only=True))
    y = (feature_df[label_col] == 'malignant').astype(int).values

    # Stage 1 — variance
    Xz = StandardScaler().fit_transform(X)
    vt = VarianceThreshold(threshold=var_thresh)
    keep_mask = vt.fit(Xz).get_support()
    X = X.loc[:, keep_mask]
    print(f'  after variance filter: {X.shape[1]} features')

    # Stage 2 correlation pruning (greedy)
    corr = X.corr().abs()
    upper = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))
    to_drop = [c for c in upper.columns if (upper[c] > corr_thresh).any()]
    X = X.drop(columns=to_drop)
    print(f'  after correlation filter (|r|>{corr_thresh}): {X.shape[1]} features')

    # Stage 3 LASSO ranking
    Xz = StandardScaler().fit_transform(X)
    clf = LogisticRegression(penalty='l1', C=lasso_C, solver='liblinear',
                             class_weight='balanced', max_iter=2000)
    clf.fit(Xz, y)
    importance = pd.Series(np.abs(clf.coef_[0]), index=X.columns)
    selected = importance.sort_values(ascending=False).head(target_k)
    print(f'  after LASSO (top-{target_k}): {len(selected)} features')

    return selected.index.tolist(), selected


# Example usage (uncomment to run after section 9's pipeline produces feature_df):
# selected_names, importance = select_stable_features(feature_df, target_k=40)
# print(importance.head(15))
# feature_df_sel = feature_df[['class','case_id','augmentation', *selected_names]]
# feature_df_sel.to_csv(output_path / 'busi_features_selected.csv', index=False)
```

## Serialize for LLM input

```python
# "BI-RADS lexicon → structured text" feeding the LLM backbone.
#
# Two outputs per case:
#   features_text      : structured BI-RADS-style report (LLM readable)
#   features_json_compact : token-efficient numeric dump (for fine-tuning / embedding)
# The text output uses real BI-RADS descriptor language, not just feature names —

def _classify_shape(feats):
    sph = feats.get('original_shape2D_Sphericity')
    elo = feats.get('original_shape2D_Elongation')
    if sph is None or elo is None:
        return None
    if sph >= 0.85 and 0.85 <= elo <= 1.0:
        return 'round'
    if sph >= 0.75:
        return 'oval'
    return 'irregular'


def _classify_orientation(feats):
    return feats.get('orientation_label')


def _classify_margin(feats):
    sph = feats.get('original_shape2D_Sphericity', 0.5)
    contrast = feats.get('original_glcm_Contrast', 0)
    
    if sph >= 0.85 and contrast < np.percentile(list(MARGIN_REF['contrast']) or [contrast], 60):
        return 'circumscribed'
    if sph < 0.6:
        return 'spiculated_or_indistinct'
    return 'not_circumscribed'


def _classify_echo(feats):
    mean = feats.get('original_firstorder_Mean')
    entropy = feats.get('original_firstorder_Entropy', 0)
    if mean is None:
        return None
    if mean < 30:
        base = 'anechoic_or_hypoechoic'
    elif mean < 80:
        base = 'hypoechoic'
    elif mean < 130:
        base = 'isoechoic'
    else:
        base = 'hyperechoic'
    if entropy > 5.5:
        base += '__heterogeneous'
    return base


def _classify_posterior(feats):
    return feats.get('posterior_label')


MARGIN_REF = {'contrast': []}


def to_llm_text(features, top_k=4, include_numeric=True):
    """Produce a BI-RADS-styled, LLM-friendly report from the radiomic feature dict."""
    grouped = categorize_lexicon(features)
    descriptors = {
        'Shape':              _classify_shape(features),
        'Orientation':        _classify_orientation(features),
        'Margin':             _classify_margin(features),
        'Echo pattern':       _classify_echo(features),
        'Posterior features': _classify_posterior(features),
    }

    lines = ['# Breast US lesion — BI-RADS-aligned radiomic profile', '']
    lines.append('## BI-RADS descriptors (derived from radiomic features)')
    for k, v in descriptors.items():
        lines.append(f'- **{k}**: {v if v is not None else "indeterminate"}')

    if include_numeric:
        lines.append('')
        lines.append('## Supporting radiomic measurements')
        for group, spec in BIRADS_LEXICON.items():
            items = {k: v for k, v in grouped.get(group, {}).items()
                     if isinstance(v, (int, float))}
            if not items:
                continue
            lines.append(f'### {spec["descriptor"]}')
            top = sorted(items.items(),
                         key=lambda kv: abs(kv[1]) if np.isfinite(kv[1]) else 0,
                         reverse=True)[:top_k]
            for name, val in top:
                short = name[len('original_'):] if name.startswith('original_') else name
                lines.append(f'  - {short}: {val:.4g}')

    return '\n'.join(lines)


def to_compact_json(features, precision=4):
    out = {}
    for k, v in features.items():
        if isinstance(v, (int, float)) and np.isfinite(v):
            out[k] = round(float(v), precision)
        elif isinstance(v, str):
            out[k] = v
    return json.dumps(out, separators=(',', ':'))


print(to_llm_text(feats))
```

## full dataset

```python
from tqdm.notebook import tqdm

def run_pipeline(cases, output_path, with_augmentation=True, max_cases=None):
    ext = build_extractor()
    cases_to_run = cases[:max_cases] if max_cases else cases
    stats = {'extracted': 0, 'mask_invalid': 0, 'failed': 0}
    
    with open(output_path, 'w') as f:
        for case in tqdm(cases_to_run, desc='Cases'):
            if case['class'] == 'normal':
                continue
            try:
                img, mask = load_pair(case['image'], case['masks'])
            except Exception as e:
                stats['failed'] += 1; continue
            
            variants = list(augment(img, mask)) if with_augmentation else [(img, mask, 'original')]
            for v_img, v_mask, tag in variants:
                ok, info = validate_mask(v_img, v_mask)
                if not ok:
                    stats['mask_invalid'] += 1; continue
                try:
                    features = extract_features(v_img, v_mask, ext)
                except Exception:
                    stats['failed'] += 1; continue
                
                record = {
                    'case_id': case['case_id'], 'class': case['class'],
                    'augmentation': tag, 'mask_info': info,
                    'n_features': len(features), 'features': features,
                    'features_text': to_llm_text(features),
                    'features_json_compact': to_compact_json(features),
                }
                f.write(json.dumps(record) + '\n')
                stats['extracted'] += 1
    return stats

# Quick test on first 5 cases (no augmentation) — sanity check before full run
out_test = output_path / 'test_5cases.jsonl'
stats = run_pipeline(cases, out_test, with_augmentation=False, max_cases=5)
print(stats)
```

```python
# Full run on all cases with augmentation
out_full = output_path / 'busi_features.jsonl'
stats = run_pipeline(cases, out_full, with_augmentation=True)
print(stats)
print(f'Output: {out_full} ({out_full.stat().st_size / 1e6:.1f} MB)')
```
