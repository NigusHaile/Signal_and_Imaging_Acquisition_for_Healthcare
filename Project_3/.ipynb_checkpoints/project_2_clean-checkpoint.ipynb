# BUSI → PyRadiomics → LLM-ready Features

Pipeline: load BUSI → validate masks → augment → extract radiomic features → serialize for LLM input.

Imports

```python
import json, logging, random
from pathlib import Path

import numpy as np
import pandas as pd
import SimpleITK as sitk
import matplotlib.pyplot as plt
from PIL import Image

import radiomics
from radiomics import featureextractor
```

```python
radiomics.logger.setLevel(logging.ERROR)

busi_dataset = Path('/media/nigus-haile/Local/EDU/milano_bicocca/2nd_semester/Healthcare/P3/BUSI_dataset')
output_path = Path('./outputs'); output_path.mkdir(exist_ok=True)

print(f'BUSI dir exists: {busi_dataset.exists()}')
print(f'swubfolders: {[p.name for p in busi_dataset.iterdir() if p.is_dir()]}')
```

## Discover cases

```python
def list_cases(busi_dir):
    """Return list of {image, masks, class, case_id} dicts.
    
    Handles BUSI structure with separate image/mask folders:
        BUSI_dataset/benign/         ← images
        BUSI_dataset/benign_mask/    ← masks
        BUSI_dataset/malignant/
        BUSI_dataset/malignant_mask/
    """
    cases = []
    img_dirs = [d for d in busi_dir.iterdir() 
                if d.is_dir() and not d.name.endswith('_mask')]
    
    for img_dir in img_dirs:
        cls = img_dir.name
        mask_dir = busi_dir / f'{cls}_mask'
        if not mask_dir.exists():
            continue
        
        for img in sorted(img_dir.glob('*.png')):
            masks = sorted(mask_dir.glob(f'{img.stem}*.png'))
            if masks:
                cases.append({
                    'image': img, 'masks': masks,
                    'class': cls, 'case_id': img.stem
                })
    return cases

cases = list_cases(busi_dataset)
print(f'Total cases: {len(cases)}')

if cases:
    df = pd.DataFrame([{'class': c['class'], 'case_id': c['case_id'],
                        'n_masks': len(c['masks'])} for c in cases])
    print(df['class'].value_counts())
else:
    print('⚠ No cases found. Check filename matching between image and mask folders.')
```

```python
for cls in ['benign', 'malignant']:
    img_dir = busi_dataset / cls
    mask_dir = busi_dataset / f'{cls}_mask'
    print(f'{cls}: {len(list(img_dir.glob("*.png")))} images, '
          f'{len(list(mask_dir.glob("*.png")))} masks')
    print(f'  sample image: {sorted(img_dir.glob("*.png"))[0].name}')
    print(f'  sample mask:  {sorted(mask_dir.glob("*.png"))[0].name}')
```

## 3. Load image + mask pair (with multi-mask merging)

```python
def load_pair(image_path, mask_paths):
    """Load image and merge multiple masks via logical OR."""
    img = np.array(Image.open(image_path).convert('L'), dtype=np.float32)
    merged = None
    for mp in mask_paths:
        m = (np.array(Image.open(mp).convert('L')) > 127).astype(np.uint8)
        merged = m if merged is None else (merged | m)
    return img, merged.astype(np.uint8)

# Show one example from each class
fig, axes = plt.subplots(2, 3, figsize=(11, 7))
for col, cls in enumerate(['benign', 'malignant', 'normal']):
    case = next((c for c in cases if c['class'] == cls), None)
    if not case: continue
    img, mask = load_pair(case['image'], case['masks'])
    axes[0, col].imshow(img, cmap='gray'); axes[0, col].set_title(f'{cls}: {case["case_id"]}')
    axes[1, col].imshow(mask, cmap='gray'); axes[1, col].set_title(f'mask ({mask.sum()} px)')
    for ax in axes[:, col]: ax.axis('off')
plt.tight_layout(); plt.show()
```

## 4. Mask validation

```python
def validate_mask(image, mask, min_voxels=100, max_area_frac=0.5):
    """Quick QC. Returns (ok: bool, info: dict)."""
    info = {'voxels': int(mask.sum()), 'area_frac': float(mask.sum() / mask.size)}
    if image.shape != mask.shape:
        return False, {**info, 'reason': f'shape mismatch {image.shape} vs {mask.shape}'}
    if info['voxels'] == 0:
        return False, {**info, 'reason': 'empty mask'}
    if info['voxels'] < min_voxels:
        return False, {**info, 'reason': f'too small ({info["voxels"]} < {min_voxels})'}
    if info['area_frac'] > max_area_frac:
        return False, {**info, 'reason': f'too large ({info["area_frac"]:.0%})'}
    edge = mask[0].sum() + mask[-1].sum() + mask[:, 0].sum() + mask[:, -1].sum()
    info['edge_pixels'] = int(edge)
    return True, info

# Validate all non-normal cases
results = []
for c in cases:
    if c['class'] == 'normal': continue
    img, mask = load_pair(c['image'], c['masks'])
    ok, info = validate_mask(img, mask)
    results.append({'case_id': c['case_id'], 'class': c['class'], 'ok': ok, **info})

val_df = pd.DataFrame(results)
print(f'Valid: {val_df["ok"].sum()} / {len(val_df)}')
val_df.head(10)
```

```python
# Show rejected cases (if any)
rejected = val_df[~val_df['ok']]
print(f'Rejected: {len(rejected)}')
rejected
```

## 5. PyRadiomics-safe augmentation

Only **geometric** transforms — preserves radiomic feature validity. Avoids brightness/blur/noise/elastic (which would corrupt texture features).

```python
def _rotate(arr, angle_deg, is_mask):
    s = sitk.GetImageFromArray(arr.astype(np.float32))
    c = [(sz - 1) / 2 for sz in s.GetSize()]
    t = sitk.Euler2DTransform(); t.SetCenter(s.TransformContinuousIndexToPhysicalPoint(c))
    t.SetAngle(np.deg2rad(angle_deg))
    interp = sitk.sitkNearestNeighbor if is_mask else sitk.sitkLinear
    out = sitk.GetArrayFromImage(sitk.Resample(s, s, t, interp, 0.0, s.GetPixelID()))
    return out.astype(np.uint8) if is_mask else out

def augment(image, mask, seed=42):
    """Yields (image, mask, tag) — original + flip + small rotation + mask perturbation."""
    rng = np.random.default_rng(seed)
    yield image, mask, 'original'
    yield np.fliplr(image).copy(), np.fliplr(mask).copy(), 'hflip'
    angle = float(rng.uniform(-10, 10))
    yield _rotate(image, angle, False), _rotate(mask, angle, True), f'rot{angle:+.1f}'
    # Mask perturbation (dilate or erode by 1 px)
    sm = sitk.GetImageFromArray(mask)
    op = sitk.BinaryDilateImageFilter() if rng.random() < 0.5 else sitk.BinaryErodeImageFilter()
    op.SetKernelRadius(1); op.SetForegroundValue(1)
    perturbed = sitk.GetArrayFromImage(op.Execute(sm)).astype(np.uint8)
    yield image.copy(), perturbed, 'mask_perturb'

# Demo on one case
sample = next(c for c in cases if c['class'] == 'malignant')
img, mask = load_pair(sample['image'], sample['masks'])
variants = list(augment(img, mask))

fig, axes = plt.subplots(1, len(variants), figsize=(14, 4))
for ax, (v_img, v_mask, tag) in zip(axes, variants):
    ax.imshow(v_img, cmap='gray')
    ax.contour(v_mask, levels=[0.5], colors='red', linewidths=1.5)
    ax.set_title(f'{tag}\n{v_mask.sum()} px'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## 6. PyRadiomics feature extractor (BI-RADS-aligned)

```python
def build_extractor():
    ext = featureextractor.RadiomicsFeatureExtractor(
        binWidth=25, normalize=True, normalizeScale=100,
        force2D=True, force2Ddimension=0, label=1,
        minimumROIDimensions=2, minimumROISize=50,
        interpolator=sitk.sitkBSpline,
    )
    ext.enableAllFeatures()
    ext.enableImageTypeByName('Original')
    ext.enableImageTypeByName('Wavelet')
    ext.enableImageTypeByName('LoG', customArgs={'sigma': [1.0, 2.0, 3.0]})
    ext.enableImageTypeByName('Gradient')
    ext.enableImageTypeByName('Square')
    return ext

def extract_features(image, mask, ext):
    s_img = sitk.GetImageFromArray(image.astype(np.float32))
    s_mask = sitk.GetImageFromArray(mask.astype(np.uint8))
    s_mask.CopyInformation(s_img)
    result = ext.execute(s_img, s_mask)
    return {k: float(v) for k, v in result.items()
            if not k.startswith('diagnostics_')
            and isinstance(v, (int, float, np.number))}

extractor = build_extractor()
feats = extract_features(img, mask, extractor)
print(f'Extracted {len(feats)} features')
pd.Series(feats).head(10)
```

## 7. Group features by BI-RADS-relevant category

```python
BIRADS_GROUPS = {
    'shape':         ['shape_'],        # BI-RADS: oval / round / irregular
    'intensity':     ['firstorder_'],   # BI-RADS: echo pattern
    'texture_glcm':  ['glcm_'],         # heterogeneity, margin sharpness
    'texture_glrlm': ['glrlm_'],
    'texture_glszm': ['glszm_'],        # microlobulation, calcifications
    'texture_gldm':  ['gldm_'],
    'texture_ngtdm': ['ngtdm_'],
    'wavelet':       ['wavelet'],       # multi-scale margin (spiculation)
    'log':           ['log-sigma'],     # edge sharpness at multiple scales
    'gradient':      ['gradient'],
    'square':        ['square_'],
}

def categorize(features):
    g = {k: {} for k in BIRADS_GROUPS}; g['other'] = {}
    for key, val in features.items():
        placed = False
        for group, prefixes in BIRADS_GROUPS.items():
            if any(p in key for p in prefixes):
                g[group][key] = val; placed = True; break
        if not placed: g['other'][key] = val
    return g

grouped = categorize(feats)
pd.Series({g: len(v) for g, v in grouped.items() if v}).sort_values(ascending=False)
```

## 8. Serialize for LLM input

```python
GROUP_LABELS = {
    'shape':         '## Shape (BI-RADS: oval / round / irregular)',
    'intensity':     '## Intensity / echo pattern',
    'texture_glcm':  '## Texture — GLCM (heterogeneity, margin sharpness)',
    'texture_glrlm': '## Texture — GLRLM (coarseness)',
    'texture_glszm': '## Texture — GLSZM (microlobulation, calcifications)',
    'wavelet':       '## Multi-scale wavelet (spiculation, margin detail)',
    'log':           '## Laplacian-of-Gaussian (edge sharpness)',
}

def to_llm_text(features, top_k=5):
    grouped = categorize(features)
    lines = ['# Lesion radiomic profile (BI-RADS aligned)']
    for group, label in GROUP_LABELS.items():
        items = grouped.get(group, {})
        if not items: continue
        top = sorted(items.items(), key=lambda kv: abs(kv[1]), reverse=True)[:top_k]
        lines.append(label)
        for name, val in top:
            short = name.split('_', 1)[-1] if '_' in name else name
            lines.append(f'  - {short}: {val:.4g}')
    return '\n'.join(lines)

def to_compact_json(features, precision=4):
    return json.dumps({k: round(float(v), precision) for k, v in features.items()},
                      separators=(',', ':'))

print(to_llm_text(feats))
```

```python
# Compact JSON preview (token-efficient)
compact = to_compact_json(feats)
print(f'JSON length: {len(compact)} chars')
print(compact[:300] + '...')
```

## 9. Run end-to-end on the full dataset

```python
from tqdm.notebook import tqdm

def run_pipeline(cases, output_path, with_augmentation=True, max_cases=None):
    ext = build_extractor()
    cases_to_run = cases[:max_cases] if max_cases else cases
    stats = {'extracted': 0, 'mask_invalid': 0, 'failed': 0}
    
    with open(output_path, 'w') as f:
        for case in tqdm(cases_to_run, desc='Cases'):
            if case['class'] == 'normal':
                continue
            try:
                img, mask = load_pair(case['image'], case['masks'])
            except Exception as e:
                stats['failed'] += 1; continue
            
            variants = list(augment(img, mask)) if with_augmentation else [(img, mask, 'original')]
            for v_img, v_mask, tag in variants:
                ok, info = validate_mask(v_img, v_mask)
                if not ok:
                    stats['mask_invalid'] += 1; continue
                try:
                    features = extract_features(v_img, v_mask, ext)
                except Exception:
                    stats['failed'] += 1; continue
                
                record = {
                    'case_id': case['case_id'], 'class': case['class'],
                    'augmentation': tag, 'mask_info': info,
                    'n_features': len(features), 'features': features,
                    'features_text': to_llm_text(features),
                    'features_json_compact': to_compact_json(features),
                }
                f.write(json.dumps(record) + '\n')
                stats['extracted'] += 1
    return stats

# Quick test on first 5 cases (no augmentation) — sanity check before full run
out_test = output_path / 'test_5cases.jsonl'
stats = run_pipeline(cases, out_test, with_augmentation=False, max_cases=5)
print(stats)
```

```python
# Full run on all cases with augmentation
out_full = OUTPUT_DIR / 'busi_features.jsonl'
stats = run_pipeline(cases, out_full, with_augmentation=True)
print(stats)
print(f'Output: {out_full} ({out_full.stat().st_size / 1e6:.1f} MB)')
```
