```python
import os
import json
import logging
import argparse
from pathlib import Path
from typing import Optional
 
import numpy as np
import SimpleITK as sitk
from PIL import Image
import radiomics
from radiomics import featureextractor

 
# Silence PyRadiomics' verbose logging — we'll do our own
radiomics.logger.setLevel(logging.ERROR)
 
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s | %(levelname)s | %(message)s',
    datefmt='%H:%M:%S'
)
log = logging.getLogger(__name__)
```

```python
# BUSI LOADER
 
class BUSILoader:
    """Loads BUSI image + mask pairs from disk.
    
    BUSI naming convention:
        <class> (<n>).png         -> grayscale ultrasound image
        <class> (<n>)_mask.png    -> binary lesion mask
        
    Some cases have multiple masks: _mask_1.png, _mask_2.png — we merge them.
    """
    
    CLASSES = ['benign', 'malignant']
    
    def __init__(self, busi_dir: str):
        self.busi_dir = Path(busi_dir)
        if not self.busi_dir.exists():
            raise FileNotFoundError(f"BUSI directory not found: {busi_dir}")
    
    def list_cases(self):
        """Return list of (image_path, mask_paths, class_label) tuples."""
        cases = []
        for cls in self.CLASSES:
            cls_dir = self.busi_dir / cls
            if not cls_dir.exists():
                log.warning(f"Class directory missing: {cls_dir}")
                continue
            
            # Find all images (files without "_mask" suffix)
            images = sorted([
                p for p in cls_dir.glob("*.png")
                if "_mask" not in p.stem
            ])
            
            for img_path in images:
                # Find corresponding mask(s) — handles multi-mask cases
                base = img_path.stem  # e.g., "benign (1)"
                masks = sorted(cls_dir.glob(f"{base}_mask*.png"))
                if not masks:
                    log.warning(f"No mask found for {img_path.name} — skipping")
                    continue
                cases.append({
                    'image_path': img_path,
                    'mask_paths': masks,
                    'class': cls,
                    'case_id': base,
                })
        
        log.info(f"Discovered {len(cases)} cases across {self.CLASSES}")
        return cases
    
    @staticmethod
    def load_pair(image_path: Path, mask_paths: list) -> tuple:
        """Load an image and merge multiple masks into one binary mask."""
        # Load image as grayscale
        img_pil = Image.open(image_path).convert('L')
        img_arr = np.array(img_pil, dtype=np.float32)
        
        # Load and merge all masks (logical OR)
        merged_mask = None
        for mp in mask_paths:
            m = np.array(Image.open(mp).convert('L'))
            m_bin = (m > 127).astype(np.uint8)
            merged_mask = m_bin if merged_mask is None else (merged_mask | m_bin)
        
        return img_arr, merged_mask.astype(np.uint8)
```

```python
# MASK VALIDATION
 
class MaskValidator:
    """PyRadiomics-aware mask quality checks.
    
    Catches the common silent failures BEFORE feature extraction runs:
      - shape mismatch image vs mask
      - non-binary mask
      - empty mask
      - lesion too small for stable texture features (<100 voxels)
      - lesion too large (likely whole-image annotation error)
      - mask touching image edge (truncated lesion)
    """
    
    MIN_VOXELS = 100        # GLCM/GLRLM need at least this for stability
    MAX_AREA_FRACTION = 0.5 # Anything larger is probably a labeling error
    
    @classmethod
    def validate(cls, image: np.ndarray, mask: np.ndarray, case_id: str = "") -> dict:
        """Returns a report dict. report['ok'] is True iff mask is usable."""
        report = {'ok': True, 'case_id': case_id, 'issues': [], 'warnings': []}
        
        # 1. Shape match
        if image.shape != mask.shape:
            report['ok'] = False
            report['issues'].append(
                f"shape mismatch: image {image.shape} vs mask {mask.shape}"
            )
            return report  # fatal — can't continue
        
        # 2. Binary
        uniques = np.unique(mask)
        if not np.array_equal(uniques, [0]) and not np.array_equal(uniques, [0, 1]):
            report['warnings'].append(f"non-binary mask values: {uniques.tolist()}")
        
        # 3. Non-empty
        n_voxels = int(mask.sum())
        report['lesion_voxels'] = n_voxels
        if n_voxels == 0:
            report['ok'] = False
            report['issues'].append("empty mask (sum=0)")
            return report
        
        # 4. Too small
        if n_voxels < cls.MIN_VOXELS:
            report['ok'] = False
            report['issues'].append(
                f"lesion too small ({n_voxels} voxels < {cls.MIN_VOXELS}); "
                "texture features will be unstable"
            )
        
        # 5. Too large (probably whole-image)
        area_frac = n_voxels / mask.size
        report['area_fraction'] = float(area_frac)
        if area_frac > cls.MAX_AREA_FRACTION:
            report['ok'] = False
            report['issues'].append(
                f"lesion fills {area_frac:.0%} of image — likely annotation error"
            )
        
        # 6. Touches image edge (truncated lesion)
        h, w = mask.shape
        edge_pixels = (
            mask[0, :].sum() + mask[-1, :].sum() +
            mask[:, 0].sum() + mask[:, -1].sum()
        )
        if edge_pixels > 0:
            report['warnings'].append(
                f"mask touches image edge ({int(edge_pixels)} pixels) — "
                "lesion may be truncated"
            )
        
        return report
```

```python
# PYRADIOMICS-SAFE AUGMENTATION
# ============================================================================
 
class RadiomicsSafeAugmenter:
    """Geometric-only augmentation that preserves PyRadiomics feature validity.
    
    AVOIDED on purpose:
      - brightness/contrast jitter (destroys first-order & GLCM features)
      - blur, sharpen, noise (destroys texture features)
      - elastic deformation (distorts shape features)
      - cutout/erase (creates fake texture inside ROI)
    
    APPLIED:
      - horizontal flip (orientation-invariant, mask follows)
      - small rotation ±10° (mask follows, nearest-neighbor for mask)
      - mask dilate/erode ±1 px (simulates annotator variability)
    """
    
    def __init__(self, rotation_max_deg: float = 10.0, do_flip: bool = True,
                 mask_perturbation: bool = True, seed: Optional[int] = None):
        self.rotation_max = rotation_max_deg
        self.do_flip = do_flip
        self.mask_perturbation = mask_perturbation
        self.rng = np.random.default_rng(seed)
    
    def generate(self, image: np.ndarray, mask: np.ndarray):
        """Yields (image, mask, augmentation_tag) tuples.
        
        Always yields the original first, then up to 3 augmentations.
        """
        # 0. Original
        yield image, mask, "original"
        
        # 1. Horizontal flip — both image and mask
        if self.do_flip:
            yield np.fliplr(image).copy(), np.fliplr(mask).copy(), "hflip"
        
        # 2. Small rotation
        angle = float(self.rng.uniform(-self.rotation_max, self.rotation_max))
        if abs(angle) > 1.0:
            img_rot = self._rotate(image, angle, is_mask=False)
            mask_rot = self._rotate(mask, angle, is_mask=True)
            yield img_rot, mask_rot, f"rot{angle:+.1f}"
        
        # 3. Mask perturbation (image unchanged) — robustness to annotator variability
        if self.mask_perturbation:
            perturbed_mask = self._perturb_mask(mask)
            if perturbed_mask.sum() > 0:
                yield image.copy(), perturbed_mask, "mask_perturb"
    
    @staticmethod
    def _rotate(arr: np.ndarray, angle_deg: float, is_mask: bool) -> np.ndarray:
        """Rotate via SimpleITK to keep image-mask alignment exact."""
        sitk_img = sitk.GetImageFromArray(arr.astype(np.float32))
        
        # Rotation around image center
        center = [(s - 1) / 2.0 for s in sitk_img.GetSize()]
        center_phys = sitk_img.TransformContinuousIndexToPhysicalPoint(center)
        
        transform = sitk.Euler2DTransform()
        transform.SetCenter(center_phys)
        transform.SetAngle(np.deg2rad(angle_deg))
        
        # CRITICAL: nearest-neighbor for masks, linear for images
        interp = sitk.sitkNearestNeighbor if is_mask else sitk.sitkLinear
        
        rotated = sitk.Resample(
            sitk_img, sitk_img, transform, interp,
            0.0,  # default pixel value
            sitk_img.GetPixelID()
        )
        out = sitk.GetArrayFromImage(rotated)
        return out.astype(np.uint8) if is_mask else out
    
    @staticmethod
    def _perturb_mask(mask: np.ndarray) -> np.ndarray:
        """Dilate or erode by 1 pixel — simulates inter-annotator variability."""
        sitk_mask = sitk.GetImageFromArray(mask)
        # Randomly pick dilate or erode
        if np.random.rand() < 0.5:
            filt = sitk.BinaryDilateImageFilter()
        else:
            filt = sitk.BinaryErodeImageFilter()
        filt.SetKernelRadius(1)
        filt.SetForegroundValue(1)
        return sitk.GetArrayFromImage(filt.Execute(sitk_mask)).astype(np.uint8)
```

```python
# PYRADIOMICS EXTRACTOR (BI-RADS-aligned settings)
# ============================================================================
 
class BIRADSRadiomicsExtractor:
    """PyRadiomics wrapper configured for breast US BI-RADS tasks.
    
    Settings rationale:
      - binWidth=25: standard for US grayscale [0,255]
      - resampling to 0.5mm isotropic: typical breast US resolution
      - bspline interpolator: smooth image resampling
      - normalize=True: critical for US (no quantitative units like CT HU)
    
    Image filters enabled:
      - Original: baseline features
      - Wavelet: captures multi-scale margin & texture (spiculation, microlobulation)
      - LoG: edge/blob detection at multiple scales (margin sharpness)
      - Gradient: boundary information
    """
    
    def __init__(self):
        settings = {
            'binWidth': 25,
            'resampledPixelSpacing': None,  # BUSI is 2D PNG — no spacing info, skip
            'interpolator': sitk.sitkBSpline,
            'normalize': True,
            'normalizeScale': 100,
            'force2D': True,
            'force2Ddimension': 0,
            'label': 1,
            'minimumROIDimensions': 2,
            'minimumROISize': 50,
        }
        
        self.extractor = featureextractor.RadiomicsFeatureExtractor(**settings)
        
        # Enable all feature classes
        self.extractor.enableAllFeatures()
        
        # Enable image filters — but skip 3D-only ones since BUSI is 2D
        self.extractor.enableImageTypeByName('Original')
        self.extractor.enableImageTypeByName('Wavelet')
        self.extractor.enableImageTypeByName('LoG', customArgs={'sigma': [1.0, 2.0, 3.0]})
        self.extractor.enableImageTypeByName('Gradient')
        self.extractor.enableImageTypeByName('Square')
    
    def extract(self, image: np.ndarray, mask: np.ndarray) -> dict:
        """Extract features. Returns numeric features only (drops diagnostics)."""
        sitk_img = sitk.GetImageFromArray(image.astype(np.float32))
        sitk_mask = sitk.GetImageFromArray(mask.astype(np.uint8))
        
        # Copy geometry so image and mask align
        sitk_mask.CopyInformation(sitk_img)
        
        result = self.extractor.execute(sitk_img, sitk_mask)
        
        # Filter out diagnostic info (keys starting with 'diagnostics_')
        features = {
            k: float(v) for k, v in result.items()
            if not k.startswith('diagnostics_') and isinstance(v, (int, float, np.number))
        }
        return features
```

```python
# LLM-READY SERIALIZATION
 
class LLMFeatureSerializer:
    """Converts numeric radiomic features into LLM-friendly representations.
    
    Produces THREE formats per case so you can pick what works best:
      1. 'features_dict' — raw numeric dict for projection-layer fusion (option 2)
      2. 'features_text' — natural-language prompt for direct LLM input (option 1)
      3. 'features_json_compact' — compact JSON string, tokenizer-friendly
    
    Also groups features by BI-RADS-relevant category so the LLM gets
    semantic structure instead of a flat number dump.
    """
    
    # Maps feature name fragments to BI-RADS-aligned groups
    BIRADS_GROUPS = {
        'shape': ['shape_'],  # → BI-RADS shape descriptors
        'intensity': ['firstorder_'],  # → echo pattern
        'texture_glcm': ['glcm_'],  # → heterogeneity, margin sharpness
        'texture_glrlm': ['glrlm_'],  # → texture coarseness
        'texture_glszm': ['glszm_'],  # → microlobulation, calcifications
        'texture_gldm': ['gldm_'],
        'texture_ngtdm': ['ngtdm_'],
        'wavelet': ['wavelet'],  # → multi-scale margin (spiculation)
        'log': ['log-sigma'],  # → margin sharpness
        'gradient': ['gradient'],
        'square': ['square_'],
    }
    
    @classmethod
    def categorize(cls, features: dict) -> dict:
        """Group features by BI-RADS-relevant category."""
        grouped = {g: {} for g in cls.BIRADS_GROUPS}
        grouped['other'] = {}
        for key, val in features.items():
            placed = False
            for group, prefixes in cls.BIRADS_GROUPS.items():
                if any(p in key for p in prefixes):
                    grouped[group][key] = val
                    placed = True
                    break
            if not placed:
                grouped['other'][key] = val
        return grouped
    
    @classmethod
    def to_natural_text(cls, features: dict, top_k_per_group: int = 5) -> str:
        """Build a structured prompt-friendly description of the lesion.
        
        Picks the most informative features per BI-RADS group (by magnitude)
        and formats them with semantic headers.
        """
        grouped = cls.categorize(features)
        
        lines = ["# Lesion radiomic profile (BI-RADS aligned)"]
        
        group_labels = {
            'shape': "## Shape (BI-RADS: oval / round / irregular)",
            'intensity': "## Intensity / echo pattern (BI-RADS: anechoic / hypo / hyper / iso)",
            'texture_glcm': "## Texture — GLCM (heterogeneity, margin sharpness)",
            'texture_glrlm': "## Texture — GLRLM (coarseness)",
            'texture_glszm': "## Texture — GLSZM (microlobulation, calcifications)",
            'wavelet': "## Multi-scale wavelet (spiculation, margin detail)",
            'log': "## Laplacian-of-Gaussian (edge sharpness at multiple scales)",
        }
        
        for group, label in group_labels.items():
            feats = grouped.get(group, {})
            if not feats:
                continue
            # Pick top-k by absolute value (most informative)
            top = sorted(feats.items(), key=lambda kv: abs(kv[1]), reverse=True)[:top_k_per_group]
            lines.append(label)
            for name, val in top:
                short_name = name.split('_', 1)[-1] if '_' in name else name
                lines.append(f"  - {short_name}: {val:.4g}")
        
        return "\n".join(lines)
    
    @classmethod
    def to_compact_json(cls, features: dict, precision: int = 4) -> str:
        """Compact JSON — token-efficient for LLM input."""
        rounded = {k: round(float(v), precision) for k, v in features.items()}
        return json.dumps(rounded, separators=(',', ':'))
```

```python
# LLM-READY SERIALIZATION
 
class LLMFeatureSerializer:
    """Converts numeric radiomic features into LLM-friendly representations.
    
    Produces THREE formats per case so you can pick what works best:
      1. 'features_dict' — raw numeric dict for projection-layer fusion (option 2)
      2. 'features_text' — natural-language prompt for direct LLM input (option 1)
      3. 'features_json_compact' — compact JSON string, tokenizer-friendly
    
    Also groups features by BI-RADS-relevant category so the LLM gets
    semantic structure instead of a flat number dump.
    """
    
    # Maps feature name fragments to BI-RADS-aligned groups
    BIRADS_GROUPS = {
        'shape': ['shape_'],  # → BI-RADS shape descriptors
        'intensity': ['firstorder_'],  # → echo pattern
        'texture_glcm': ['glcm_'],  # → heterogeneity, margin sharpness
        'texture_glrlm': ['glrlm_'],  # → texture coarseness
        'texture_glszm': ['glszm_'],  # → microlobulation, calcifications
        'texture_gldm': ['gldm_'],
        'texture_ngtdm': ['ngtdm_'],
        'wavelet': ['wavelet'],  # → multi-scale margin (spiculation)
        'log': ['log-sigma'],  # → margin sharpness
        'gradient': ['gradient'],
        'square': ['square_'],
    }
    
    @classmethod
    def categorize(cls, features: dict) -> dict:
        """Group features by BI-RADS-relevant category."""
        grouped = {g: {} for g in cls.BIRADS_GROUPS}
        grouped['other'] = {}
        for key, val in features.items():
            placed = False
            for group, prefixes in cls.BIRADS_GROUPS.items():
                if any(p in key for p in prefixes):
                    grouped[group][key] = val
                    placed = True
                    break
            if not placed:
                grouped['other'][key] = val
        return grouped
    
    @classmethod
    def to_natural_text(cls, features: dict, top_k_per_group: int = 5) -> str:
        """Build a structured prompt-friendly description of the lesion.
        
        Picks the most informative features per BI-RADS group (by magnitude)
        and formats them with semantic headers.
        """
        grouped = cls.categorize(features)
        
        lines = ["# Lesion radiomic profile (BI-RADS aligned)"]
        
        group_labels = {
            'shape': "## Shape (BI-RADS: oval / round / irregular)",
            'intensity': "## Intensity / echo pattern (BI-RADS: anechoic / hypo / hyper / iso)",
            'texture_glcm': "## Texture — GLCM (heterogeneity, margin sharpness)",
            'texture_glrlm': "## Texture — GLRLM (coarseness)",
            'texture_glszm': "## Texture — GLSZM (microlobulation, calcifications)",
            'wavelet': "## Multi-scale wavelet (spiculation, margin detail)",
            'log': "## Laplacian-of-Gaussian (edge sharpness at multiple scales)",
        }
        
        for group, label in group_labels.items():
            feats = grouped.get(group, {})
            if not feats:
                continue
            # Pick top-k by absolute value (most informative)
            top = sorted(feats.items(), key=lambda kv: abs(kv[1]), reverse=True)[:top_k_per_group]
            lines.append(label)
            for name, val in top:
                short_name = name.split('_', 1)[-1] if '_' in name else name
                lines.append(f"  - {short_name}: {val:.4g}")
        
        return "\n".join(lines)
    
    @classmethod
    def to_compact_json(cls, features: dict, precision: int = 4) -> str:
        """Compact JSON — token-efficient for LLM input."""
        rounded = {k: round(float(v), precision) for k, v in features.items()}
        return json.dumps(rounded, separators=(',', ':'))
```

```python
# END-TO-END PIPELINE

def run_pipeline(busi_dir: str, output_path: str, 
                 with_augmentation: bool = True,
                 max_cases: Optional[int] = None) -> dict:
    """Process the entire BUSI dataset end-to-end.
    
    Writes one JSON-Lines record per (case × augmentation) to `output_path`.
    Each record contains:
        - case_id, class, augmentation_tag
        - mask validation report
        - raw features dict (numeric)
        - features_text (LLM-ready natural language)
        - features_json_compact
    """
    loader = BUSILoader(busi_dir)
    validator = MaskValidator()
    augmenter = RadiomicsSafeAugmenter(seed=42) if with_augmentation else None
    extractor = BIRADSRadiomicsExtractor()
    serializer = LLMFeatureSerializer()
    
    cases = loader.list_cases()
    if max_cases:
        cases = cases[:max_cases]
    
    stats = {'total': 0, 'extracted': 0, 'mask_invalid': 0, 'extraction_failed': 0,
             'by_class': {c: 0 for c in BUSILoader.CLASSES}}
    
    with open(output_path, 'w') as out_f:
        for case_idx, case in enumerate(cases, 1):
            log.info(f"[{case_idx}/{len(cases)}] {case['case_id']} ({case['class']})")
            
            try:
                image, mask = loader.load_pair(case['image_path'], case['mask_paths'])
            except Exception as e:
                log.error(f"  load failed: {e}")
                continue
            
            # Skip "normal" cases — they have no lesion to extract features from
            if case['class'] == 'normal' or mask.sum() == 0:
                log.info("  no lesion — skipping (normal case)")
                continue
            
            # Generate augmented versions (or just the original)
            if augmenter:
                variants = list(augmenter.generate(image, mask))
            else:
                variants = [(image, mask, "original")]
            
            for var_img, var_mask, aug_tag in variants:
                stats['total'] += 1
                
                # MASK VALIDATION
                report = validator.validate(var_img, var_mask, case['case_id'])
                if not report['ok']:
                    log.warning(f"  [{aug_tag}] mask invalid: {report['issues']}")
                    stats['mask_invalid'] += 1
                    continue
                
                # FEATURE EXTRACTION
                try:
                    features = extractor.extract(var_img, var_mask)
                except Exception as e:
                    log.error(f"  [{aug_tag}] extraction failed: {e}")
                    stats['extraction_failed'] += 1
                    continue
                
                # SERIALIZATION
                record = {
                    'case_id': case['case_id'],
                    'class': case['class'],
                    'augmentation': aug_tag,
                    'mask_report': {
                        'lesion_voxels': report.get('lesion_voxels'),
                        'area_fraction': report.get('area_fraction'),
                        'warnings': report.get('warnings', []),
                    },
                    'n_features': len(features),
                    'features': features,
                    'features_text': serializer.to_natural_text(features),
                    'features_json_compact': serializer.to_compact_json(features),
                }
                out_f.write(json.dumps(record) + '\n')
                stats['extracted'] += 1
                stats['by_class'][case['class']] += 1
    
    log.info("=" * 60)
    log.info(f"Pipeline complete. Stats: {json.dumps(stats, indent=2)}")
    log.info(f"Output written to: {output_path}")
    return stats
```

```python
# CLI
 
def main():
    parser = argparse.ArgumentParser(description="BUSI -> PyRadiomics -> LLM pipeline")
    parser.add_argument('--busi-dir', required=True,
                        help="Path to Dataset_BUSI_with_GT directory")
    parser.add_argument('--output', default='busi_features.jsonl',
                        help="Output JSONL file path")
    parser.add_argument('--no-augmentation', action='store_true',
                        help="Disable augmentation (extract original only)")
    parser.add_argument('--max-cases', type=int, default=None,
                        help="Process only the first N cases (for testing)")
    args = parser.parse_args()
    
    run_pipeline(
        busi_dir=args.busi_dir,
        output_path=args.output,
        with_augmentation=not args.no_augmentation,
        max_cases=args.max_cases,
    )
 
 
if __name__ == '__main__':
    main()
 
```
